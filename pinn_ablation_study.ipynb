# PINN Ablation Study: Controlled A/B Experiment

**Date**: June 2, 2026
**Objective**: Measure the isolated impact of Physics-Informed Neural Networks (PINN) on system performance

## Experimental Design

### Experiment A (Baseline): Without PINN
- PINN: **DISABLED**
- ACO: Enabled
- Emergency Corridor: Enabled
- Predictive Green Wave: Enabled
- Traffic Clearing: Enabled

### Experiment B (Enhanced): With PINN
- PINN: **ENABLED**
- ACO: Enabled
- Emergency Corridor: Enabled
- Predictive Green Wave: Enabled
- Traffic Clearing: Enabled

### Controlled Parameters (Same for Both)
- **Simulation Duration**: 100 seconds (1000 steps)
- **Traffic Demand**: Fixed multiplier (1.0x)
- **Random Seed**: 42 (reproducible)
- **Emergency Schedule**: Same emergency vehicle triggers
- **Network Topology**: Same SUMO network file
- **Step Length**: 0.1 seconds (10Hz)

## Metrics Collected

### Traffic Flow Metrics
1. **Average Waiting Time** (seconds) - Lower is better
2. **Average Queue Length** (vehicles) - Lower is better
3. **Average Speed** (m/s) - Higher is better
4. **Throughput** (vehicles/step) - Higher is better

### Emergency Response Metrics
5. **Emergency Travel Time** (seconds) - Lower is better
6. **Emergency Vehicle Stops** (count) - Lower is better

### System Activity Metrics
7. **ACO Reroutes** (count) - Higher indicates better adaptation
8. **Corridor Activations** (count) - Higher indicates more proactive response

## Hypothesis

**H0 (Null)**: PINN provides no significant performance improvement  
**H1 (Alternative)**: PINN improves at least 5/8 metrics
